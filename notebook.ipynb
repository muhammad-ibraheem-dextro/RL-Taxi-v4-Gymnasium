# Custom Reinforcement Learning Agent (Q-Learning) for Taxi-v4
This notebook demonstrates the training and execution of a Q-Learning agent on the Gymnasium Taxi-v4 environment.

```python
import gymnasium as gym
import numpy as np
from agent import QLearningAgent
from utils import plot_training_progress, record_gif

# Training the agent directly in the notebook
env = gym.make("Taxi-v4")
state_size = env.observation_space.n
action_size = env.action_space.n

agent = QLearningAgent(state_size, action_size, learning_rate=0.1, discount_factor=0.99)
episodes = 2500
rewards_history = []

print("Training Started! Please wait...")
for ep in range(episodes):
    state, _ = env.reset()
    total_reward = 0
    done = False
    
    while not done:
        action = agent.choose_action(state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        
        agent.update(state, action, reward, next_state, done)
        state = next_state
        total_reward += reward
        
    agent.decay_epsilon()
    rewards_history.append(total_reward)
    
    if (ep + 1) % 500 == 0:
        print(f"Episode: {ep+1}/{episodes} | Avg Reward: {np.mean(rewards_history[-100:]):.2f}")

plot_training_progress(rewards_history)
env.close()
```
